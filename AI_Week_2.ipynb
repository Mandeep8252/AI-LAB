```python
class EightPuzzleDFS:
    def __init__(self, start_state, goal_state=None):
        self.start_state = tuple(tuple(row) for row in start_state)
        if goal_state is None:
            self.goal_state = ((1, 2, 3), (4, 5, 6), (7, 8, 0))
        else:
            self.goal_state = tuple(tuple(row) for row in goal_state)

        # Possible movements of the blank space (0): Up, Down, Left, Right
        self.moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    def find_blank(self, state):
        """Finds the row and column index of the blank space (0)."""
        for r in range(3):
            for c in range(3):
                if state[r][c] == 0:
                    return r, c
        return None

    def get_neighbors(self, state):
        """Generates all valid next states from the current state."""
        neighbors = []
        r, c = self.find_blank(state)

        for dr, dc in self.moves:
            nr, nc = r + dr, c + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                # Convert tuple to mutable list to swap elements
                new_board = [list(row) for row in state]
                # Swap the blank space with the target tile
                new_board[r][c], new_board[nr][nc] = new_board[nr][nc], new_board[r][c]
                # Convert back to immutable nested tuple
                neighbors.append(tuple(tuple(row) for row in new_board))
        return neighbors

    def solve(self, max_depth=50):
        """Solves the puzzle using Depth-Limited Depth-First Search (Stack-based)."""
        # Stack elements store: (current_state, path_taken, current_depth)
        stack = [(self.start_state, [self.start_state], 0)]
        visited = set()

        while stack:
            current_state, path, depth = stack.pop()

            if current_state == self.goal_state:
                return path

            if depth >= max_depth:
                continue

            if current_state not in visited:
                visited.add(current_state)

                # Get neighbors and push them onto the stack
                for neighbor in self.get_neighbors(current_state):
                    if neighbor not in visited:
                        stack.append((neighbor, path + [neighbor], depth + 1))

        return None

    def print_board(self, state):
        """Utility to format and print the 3x3 board cleanly."""
        for row in state:
            print(" ".join(str(x) if x != 0 else "_" for x in row))
        print("-" * 5)


# --- Execution Example ---
if __name__ == "__main__":
    # Define an easy-to-solve initial state (only a few moves away)
    # Note: Many random configurations require deep paths that standard DFS cannot easily find.
    initial_board = [
        [1, 2, 3],
        [4, 0, 6],
        [7, 5, 8]
    ]

    solver = EightPuzzleDFS(initial_board)
    print("Initial State:")
    solver.print_board(solver.start_state)

    # Solve with a maximum depth limit of 20
    solution_path = solver.solve(max_depth=20)

    if solution_path:
        print(f"Solution found in {len(solution_path) - 1} steps!\n")
        for step, state in enumerate(solution_path):
            print(f"Step {step}:")
            solver.print_board(state)
    else:
        print("No solution found within the specified depth limit.")
```

```python

def get_neighbors(state):
    neighbors = []

    index = state.index(0)
    row = index // 3
    col = index % 3

    moves = [
        (-1, 0), # Up
        (1, 0), # Down
        (0, -1), # Left
        (0, 1) # Right
    ]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_index = new_row * 3 + new_col

            new_state = list(state)
            new_state[index], new_state[new_index] = \
                new_state[new_index], new_state[index]

            neighbors.append(tuple(new_state))

    return neighbors


def depth_limited_search(state, goal, depth, path):

    if state == goal:
        return path

    if depth == 0:
        return None

    for next_state in get_neighbors(state):

        if next_state not in path:
            result = depth_limited_search(
                next_state,
                goal,
                depth - 1,
                path + [next_state]
            )

            if result:
                return result

    return None


def idf(start, goal):

    depth = 0

    while True:
        result = depth_limited_search(
            start,
            goal,
            depth,
            [start]
        )

        if result:
            return result

        depth += 1


def print_puzzle(state):

    for i in range(0, 9, 3):
        print(state[i:i+3])

    print()


# Initial state
start = (
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
)

# Goal state
goal = (
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
)


# Solve the puzzle
solution = idf(start, goal)


print("Solution Found!")
print("Number of moves:", len(solution) - 1)
print()

for step in solution:
    print_puzzle(step)
```
