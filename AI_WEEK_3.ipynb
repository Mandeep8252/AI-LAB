```python
import random
import math

# ============================================================
# 8 QUEENS - SIMULATED ANNEALING
# ============================================================

N = 8

# Initial temperature
INITIAL_TEMPERATURE = 10.0

# Cooling factor
ALPHA = 0.90

# For reproducible output
random.seed(0)


# ============================================================
# COST FUNCTION
# ============================================================

def calculate_cost(board):

    cost = 0

    for c1 in range(N):
        for c2 in range(c1 + 1, N):

            r1 = board[c1]
            r2 = board[c2]

            # Same row
            if r1 == r2:
                cost += 1

            # Same diagonal
            elif abs(r1 - r2) == abs(c1 - c2):
                cost += 1

    return cost


# ============================================================
# GENERATE NEIGHBOUR
# ============================================================

def generate_neighbour(board):

    new_board = board.copy()

    # Pick random column
    column = random.randint(0, 7)

    # Current row
    old_row = new_board[column]

    # Pick a different random row
    new_row = random.randint(0, 7)

    while new_row == old_row:
        new_row = random.randint(0, 7)

    new_board[column] = new_row

    return new_board, column, old_row, new_row


# ============================================================
# PRINT CHESS BOARD
# ============================================================

def print_board(board):

    for row in range(8):

        for col in range(8):

            if board[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")

        print()


# ============================================================
# SIMULATED ANNEALING
# ============================================================

def simulated_annealing():

    # --------------------------------------------------------
    # Initial configuration from PDF
    # --------------------------------------------------------

    current = [0, 1, 2, 3, 4, 5, 6, 7]

    current_cost = calculate_cost(current)

    T = INITIAL_TEMPERATURE

    iteration = 1
    restart = 0

    print("=" * 70)
    print("             8-QUEENS SIMULATED ANNEALING")
    print("=" * 70)

    print("\nINITIAL CONFIGURATION")
    print("-" * 70)

    print("Temperature :", T)
    print("Alpha       :", ALPHA)
    print("Board       :", current)
    print("Cost        :", current_cost)

    print("\nBoard:")
    print_board(current)

    # --------------------------------------------------------
    # Continue until cost becomes 0
    # --------------------------------------------------------

    while current_cost != 0:

        # ====================================================
        # Generate neighbour
        # ====================================================

        next_board, column, old_row, new_row = generate_neighbour(current)

        next_cost = calculate_cost(next_board)

        # PDF:
        # ΔE = Cost_new - Cost_current

        delta_E = next_cost - current_cost

        print("\n")
        print("=" * 70)
        print("ITERATION :", iteration)
        print("=" * 70)

        print("Temperature :", round(T, 6))

        # ====================================================
        # CURRENT STATE
        # ====================================================

        print("\nCURRENT STATE")
        print("-" * 30)

        print("Board :", current)
        print("Cost  :", current_cost)

        print_board(current)

        # ====================================================
        # PROPOSED MOVE
        # ====================================================

        print("\nPROPOSED MOVE")
        print("-" * 30)

        print("Column :", column)
        print("Old Row:", old_row)
        print("New Row:", new_row)

        print("\nNext Board :", next_board)
        print("Next Cost  :", next_cost)

        print_board(next_board)

        # ====================================================
        # DELTA E
        # ====================================================

        print("\nΔE = New Cost - Current Cost")
        print("ΔE =", next_cost, "-", current_cost, "=", delta_E)

        # ====================================================
        # ACCEPTANCE
        # ====================================================

        if delta_E < 0:

            # Better solution
            # ALWAYS ACCEPT

            current = next_board
            current_cost = next_cost

            print("\nDECISION : ACCEPT")
            print("Reason   : ΔE < 0 → Better solution")

        else:

            # Worse/equal solution
            # P = e^(-ΔE/T)

            probability = math.exp(-delta_E / T)

            R = random.random()

            print("\nAcceptance Probability")

            print("P = e^(-ΔE / T)")
            print(
                "P = e^(-{} / {})".format(
                    delta_E,
                    round(T, 6)
                )
            )

            print("P =", round(probability, 6))

            print("\nRandom Number R =", round(R, 6))

            if R < probability:

                current = next_board
                current_cost = next_cost

                print("\nDECISION : ACCEPT")
                print("Reason   : R < P")

            else:

                print("\nDECISION : REJECT")
                print("Reason   : R >= P")
                print("Current board remains unchanged.")

        # ====================================================
        # CHECK SOLUTION
        # ====================================================

        if current_cost == 0:

            print("\n")
            print("#" * 70)
            print("                 SOLUTION FOUND!")
            print("#" * 70)

            print("\nFinal Board :", current)
            print("Final Cost  :", current_cost)
            print("Iterations  :", iteration)

            print("\nFINAL 8-QUEENS BOARD")
            print("-" * 30)

            print_board(current)

            break

        # ====================================================
        # COOLING
        # ====================================================

        old_temperature = T

        T = T * ALPHA

        print("\nTEMPERATURE UPDATE")
        print("-" * 30)

        print(
            "T = {:.6f} × {:.2f} = {:.6f}".format(
                old_temperature,
                ALPHA,
                T
            )
        )

        # ====================================================
        # REHEATING
        # ====================================================

        # If temperature becomes extremely small,
        # restart temperature so SA can continue searching.

        if T < 0.01:

            restart += 1

            print("\n*** TEMPERATURE VERY LOW ***")
            print("Reheating Simulated Annealing...")
            print("Restart number:", restart)

            T = INITIAL_TEMPERATURE

            print("Temperature reset to:", T)

        iteration += 1


# ============================================================
# RUN PROGRAM
# ============================================================

simulated_annealing()
```
